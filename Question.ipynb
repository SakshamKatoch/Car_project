Q1. What is the average price of cars?

```python
import pandas as pd

df = pd.read_csv("cars.csv")

df['AskPrice'] = df['AskPrice'].replace('[₹,]', '', regex=True)

df['AskPrice'] = pd.to_numeric(df['AskPrice'], errors='coerce')

average_price = df['AskPrice'].mean()

print("Average Price of Cars: ₹", round(average_price, 2))
```
